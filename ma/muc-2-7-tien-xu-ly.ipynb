## 2.7. Thực hành

Bài thực hành tổng hợp các kỹ thuật tiền xử lý trên dữ liệu sinh viên. Chạy lần lượt các bước, ghi lại kết quả và đối chiếu bằng cùng tệp dữ liệu, thứ tự xử lý và phiên bản thư viện.

Thực hiện phần chuẩn bị, sau đó lần lượt các Phần A đến E. A làm sạch; B tích hợp, biến đổi và xây dựng đặc trưng; C rút gọn chiều dữ liệu và lựa chọn đặc trưng; D chia tập và cân bằng; E tạo quy trình tiền xử lý. A–C minh họa thao tác trên toàn bộ dữ liệu, không dùng để báo cáo hiệu quả dự báo. Khi đánh giá mô hình, chia tập trước và chỉ khớp các phép tiền xử lý trên tập huấn luyện như Phần E.

### Phần A: Áp dụng các kỹ thuật làm sạch dữ liệu (Mục 2.3)

### Mục tiêu

Áp dụng ba kỹ thuật làm sạch đã học ở Mục 2.3:

- (a) Xử lý dữ liệu thiếu trong cột Tuoi (Mục 2.3.1)

- (b) Phát hiện và loại bỏ ngoại lai bằng IQR (Mục 2.3.2)

- (c) Làm trơn nhiễu bằng phân khoảng (Mục 2.3.3)



````python
# Phan A.1: Xu ly du lieu thieu (Muc 2.3.1)
print('So gia tri thieu trong cot Tuoi:',
      df['Tuoi'].isnull().sum())

median_tuoi = df['Tuoi'].median()
print(f'Trung vi cot Tuoi: {median_tuoi}')
df['Tuoi'] = df['Tuoi'].fillna(median_tuoi)
print('Sau khi dien, so gia tri thieu:',
      df['Tuoi'].isnull().sum())
````

### Đối chiếu kết quả A.1

**Kết quả đối chiếu**

```
So gia tri thieu trong cot Tuoi: 12. Trung vi cot Tuoi: 20.0. Sau khi dien, so gia tri thieu: 0.
```



````python
# Phan A.2: Phat hien ngoai le bang IQR (Muc 2.3.2)
Q1 = df['Tuoi'].quantile(0.25)
Q3 = df['Tuoi'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print(f'Q1 = {Q1}, Q3 = {Q3}, IQR = {IQR}')
print(f'Can duoi = {lower}, can tren = {upper}')

outliers = df[(df['Tuoi'] < lower) | (df['Tuoi'] > upper)]
print(f'So ngoai le phat hien: {len(outliers)}')
print('Cac gia tri ngoai le:',
      sorted(outliers['Tuoi'].unique().tolist()))

# Loai bo ngoai le
df = df[(df['Tuoi'] >= lower) & (df['Tuoi'] <= upper)].copy()
print(f'So ban ghi sau khi loai ngoai le: {len(df)}')
````

### Đối chiếu kết quả A.2

**Kết quả đối chiếu**

```
Q1 = 19.0, Q3 = 22.0, IQR = 3.0. Can duoi = 14.5, can tren = 26.5.
```

**Kết quả đối chiếu**

```
So ngoai le phat hien: 3
```

**Kết quả đối chiếu**

```
Cac gia tri ngoai le: [5.0, 99.0, 150.0]
```

**Kết quả đối chiếu**

```
So ban ghi sau khi loai ngoai le: 197. Ba giá trị 5, 99 và 150 là lỗi nhập liệu được cố ý đưa vào để minh họa IQR
```



````python
# Phan A.3: Lam tron nhieu bang phan khoang (Muc 2.3.3)
df['DiemTB_bin'] = pd.cut(df['DiemTB_HK1'],
                           bins=4, labels=False)
bin_means = df.groupby('DiemTB_bin')['DiemTB_HK1'].mean()
print('Trung binh moi bin:')
print(bin_means)

df['DiemTB_HK1_smooth'] = df['DiemTB_bin'].map(bin_means)
print(df[['DiemTB_HK1', 'DiemTB_HK1_smooth']].head())
````

### Đối chiếu kết quả A.3

**Kết quả đối chiếu**

```

Bốn khoảng rộng bằng nhau trên đoạn từ 1,518 đến 4,0; vì điểm trung bình không phân bố đều nên số mẫu mỗi khoảng khác nhau. Mỗi giá trị DiemTB_HK1 được thay bằng trung bình của khoảng chứa nó, nên cột DiemTB_HK1_smooth chỉ còn bốn giá trị khác nhau.

### Phần B: Áp dụng các kỹ thuật tích hợp và biến đổi (Mục 2.4)

### Mục tiêu

Áp dụng các kỹ thuật của Mục 2.4:

- (a) Tích hợp dữ liệu từ một nguồn phụ giả định (Mục 2.4.1)

- (b) Biến đổi: chuẩn hóa các cột số và mã hóa cột phân loại (Mục 2.4.2)

- (c) Xây dựng đặc trưng mới (Mục 2.4.2)



````python
# Phan B.1: Tich hop du lieu tu nguon phu (Muc 2.4.1)
# Gia su co bang phu chua thong tin hoat dong LMS
rng = np.random.default_rng(20260518)  # Cung seed de tai lap ket qua
extra = pd.DataFrame({'MaSV': df['MaSV'].values,
    'SoLanDangNhapLMS': rng.integers(5, 50, len(df))})
df = df.merge(extra, on='MaSV', how='left')
print(f'Sau khi tich hop: {df.shape}')
````

### Đối chiếu kết quả B.1

**Kết quả đối chiếu**

```
Sau khi tich hop: (197, 11). Số bản ghi giữ nguyên 197 vì MaSV là duy nhất - đây là điều kiện bắt buộc khi merge; nếu khóa ghép bị trùng, số dòng sẽ tăng lên ngoài ý muốn. Số cột tăng từ 10 lên 11 do thêm cột SoLanDangNhapLMS.
```



````python
# Phan B.2: Bien doi du lieu (Muc 2.4.2)
from sklearn.preprocessing import StandardScaler

num_cols = ['Tuoi', 'DiemTB_HK1', 'SoTinChi',
            'SoBuoiVangMat', 'SoLanDangNhapLMS']
scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])

print('Sau khi chuan hoa:')
print('Trung binh:', df[num_cols].mean().round(6).to_dict())
print('Do lech chuan:', df[num_cols].std().round(3).to_dict())

# One-Hot Encoding cho cot phan loai
df = pd.get_dummies(df, columns=['NganhHoc', 'HinhThucDangKy'],
                     drop_first=True)
print(f'Sau One-Hot Encoding: {df.shape[1]} cot')
````

### Đối chiếu kết quả B.2

**Kết quả đối chiếu**

```
sau StandardScaler, trung bình mỗi cột số xấp xỉ 0 và độ lệch chuẩn mẫu xấp xỉ 1.003 (hệ số 1.003 đến từ ddof = 1 trên 197 bản ghi). Sau One-Hot Encoding: 12 cột. Cột tăng từ 11 lên 12 vì NganhHoc (3 giá trị) thành 2 cột và HinhThucDangKy (2 giá trị) thành 1 cột, đồng thời bỏ 2 cột gốc.
```



````python
# Phan B.3: Xay dung dac trung (Muc 2.4.2)
# Luu y: dung ban truoc chuan hoa de tinh dac trung moi
# (vi vay can chay phan nay TRUOC khi chuan hoa
# hoac trich lai gia tri tu file goc)

# O day minh hoa cong thuc tao dac trung TyLeThamGia
# tinh tu du lieu goc:
df_goc = pd.read_csv('datasets/raw/dmml_sinhvien.csv')
median_goc = df_goc['Tuoi'].median()
df_goc['Tuoi'] = df_goc['Tuoi'].fillna(median_goc)
Q1g = df_goc['Tuoi'].quantile(0.25)
Q3g = df_goc['Tuoi'].quantile(0.75)
IQRg = Q3g - Q1g
df_goc = df_goc[(df_goc['Tuoi'] >= Q1g - 1.5*IQRg)
                & (df_goc['Tuoi'] <= Q3g + 1.5*IQRg)]
df_goc['TyLeThamGia'] = (1 - df_goc['SoBuoiVangMat']
                         / (df_goc['SoTinChi'] * 3))
print(df_goc['TyLeThamGia'].describe())
````

### Đối chiếu kết quả B.3

**Kết quả đối chiếu**

count 197, mean 0.9608, std 0.0256, min 0.8889, 50% 0.9630, max 1.0000. Công thức giả định mỗi tín chỉ tương ứng 3 buổi học. Giá trị 1 nghĩa là không vắng buổi nào; giá trị thấp hơn ứng với tỷ lệ vắng cao hơn.

### Phần C: Áp dụng các kỹ thuật rút gọn và lựa chọn đặc trưng (Mục 2.5)

### Mục tiêu

Áp dụng các kỹ thuật của Mục 2.5:

- (a) Rút gọn chiều dữ liệu bằng PCA (Mục 2.5.1)

- (b) Lựa chọn đặc trưng bằng SelectKBest (Mục 2.5.2)



````python
# Phan C: Tiep tuc tu Phan B (df sau B.2)
from sklearn.decomposition import PCA
from sklearn.feature_selection import SelectKBest, f_classif

# Tach features va target
drop_cols = ['MaSV', 'KetQua', 'DiemTB_bin', 'DiemTB_HK1_smooth']
X = df.drop(columns=[c for c in drop_cols if c in df.columns])
y = df['KetQua']
print(f'Kich thuoc X: {X.shape}')

# Phan C.1: PCA
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)
print('Ty le phuong sai 3 thanh phan chinh:',
      pca.explained_variance_ratio_.round(4))
print('Tong:',
      round(pca.explained_variance_ratio_.sum(), 4))
````

**Kết quả đối chiếu**

```
Kich thuoc X: (197, 8)
```

**Kết quả đối chiếu**

```

```

**Kết quả đối chiếu**

```
Tong: 0.5885. Ba thành phần đầu giữ 58,85% phương sai. Muốn giữ ít nhất 80% thì cần 5 thành phần, vì phương sai tích lũy lần lượt là 21,15%; 41,16%; 58,85%; 74,77% và 88,40%. Tỷ lệ phương sai không cho biết các đặc trưng gốc có độc lập với nhau hay không. Lưu ý rằng Phần C tiếp nối Phần B: các cột số đã được chuẩn hóa ở B.2 nên không chuẩn hóa lại, còn các cột one-hot chỉ nhận giá trị 0 hoặc 1 và được giữ nguyên. Vì vậy Bước 1 của Mục 2.5.1 không được lặp lại ở đây, và tỷ lệ phương sai thu được chỉ so sánh được với lần chạy theo đúng thứ tự này.

````python
# Phan C.2: SelectKBest
selector = SelectKBest(score_func=f_classif, k=5)
X_selected = selector.fit_transform(X, y)

selected_cols = X.columns[selector.get_support()]
print('5 dac trung quan trong nhat:')
print(list(selected_cols))

# Xem diem F cua tat ca dac trung
import pandas as pd
scores_df = pd.DataFrame({
    'feature': X.columns,
    'F_score': selector.scores_
}).sort_values('F_score', ascending=False)
print(scores_df)
````

### Phần D: Áp dụng các kỹ thuật lấy mẫu dữ liệu (Mục 2.6)

### Mục tiêu

Áp dụng các kỹ thuật của Mục 2.6:

- (a) Chia huấn luyện/kiểm tra bằng lấy mẫu phân tầng (Mục 2.6.1)

- (b) Cân bằng dữ liệu bằng Random Oversampling chỉ trên tập huấn luyện (Mục 2.6.3)



````python
# Phan D.1: Chia train/test voi stratified sampling
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print(f'Tap train: {len(y_train)} ban ghi')
print('Phan bo train:', y_train.value_counts().to_dict())
print(f'Tap test: {len(y_test)} ban ghi')
print('Phan bo test:', y_test.value_counts().to_dict())
````

### Đối chiếu kết quả D.1

**Kết quả đối chiếu**

```
Tap train: 157 ban ghi,
```

**Kết quả đối chiếu**

```
Phan bo train: {'Dat': 133, 'Khong dat': 24}.
```

**Kết quả đối chiếu**

```
Tap test: 40 ban ghi,
```

**Kết quả đối chiếu**

```
Phan bo test: {'Dat': 34, 'Khong dat': 6}. Tỷ lệ Đạt trên Không đạt giữ nguyên khoảng 85/15 ở cả hai tập nhờ tham số stratify=y.
```



````python
# Phan D.2: lap mau, giu nguyen cac gia tri one-hot
from imblearn.over_sampling import RandomOverSampler
print('Truoc lay mau:', y_train.value_counts().to_dict())
ros = RandomOverSampler(random_state=42)
X_train_balanced, y_train_balanced = ros.fit_resample(X_train, y_train)
print('Sau lay mau:',
      pd.Series(y_train_balanced).value_counts().to_dict())
print('Tong so ban ghi train:', len(y_train_balanced))
````

```

Dữ liệu Phần D kế thừa các thao tác minh họa trên toàn bộ dữ liệu ở A–C, nên không dùng kết quả này để báo cáo hiệu năng dự báo. Khi đánh giá, dùng quy trình chia tập trước như Phần E và imblearn.pipeline.Pipeline nếu có bước lấy mẫu. Với dữ liệu hỗn hợp cần sinh mẫu mới, có thể dùng SMOTENC trước mã hóa one-hot; không dùng SMOTE thông thường để nội suy các cột phân loại.

### Mục tiêu

Kết hợp điền thiếu, chuẩn hóa, mã hóa, chọn đặc trưng và PCA vào một pipeline minh họa bằng scikit-learn. Ví dụ này chưa bao gồm loại ngoại lai, tích hợp bảng LMS, tạo đặc trưng mới hay cân bằng lớp. Pipeline có ba lợi ích chính:

- (b) Đảm bảo các bước được áp dụng đúng thứ tự

- (c) Dễ tái sử dụng và triển khai trong môi trường sản xuất



````python
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

# Doc lai dataset goc
df = pd.read_csv('datasets/raw/dmml_sinhvien.csv')
X_raw = df.drop(columns=['MaSV', 'KetQua'])
y_raw = df['KetQua']

# Pipeline cho cot so
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Pipeline cho cot phan loai
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipeline,
     ['Tuoi', 'DiemTB_HK1', 'SoTinChi', 'SoBuoiVangMat']),
    ('cat', categorical_pipeline,
     ['NganhHoc', 'HinhThucDangKy'])
])

full_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('selector', SelectKBest(score_func=f_classif, k=5)),
    ('pca', PCA(n_components=3))
])

X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.2, stratify=y_raw, random_state=42)

X_transformed = full_pipeline.fit_transform(X_train, y_train)
print(f'Du lieu sau pipeline: {X_transformed.shape}')
````

Mục 2.5 trình bày lựa chọn đặc trưng và rút gọn chiều dữ liệu như hai cách tiếp cận riêng; ở đây hai bước được xếp nối tiếp chỉ để minh họa cách khai báo nhiều bước trong một pipeline, không phải khuyến nghị mặc định cho mọi bài toán. Số bản ghi 160 là tập huấn luyện sau khi chia (đúng 80% của 200 bản ghi gốc - pipeline không loại ngoại lai nên giữ đủ 200 bản ghi trước khi chia). Hình 2.11 tách rõ hai nhánh chạy song song bên trong ColumnTransformer: nhánh cột số đi qua SimpleImputer với trung vị rồi StandardScaler, nhánh cột phân loại đi qua SimpleImputer với giá trị hay gặp nhất rồi OneHotEncoder, hai nhánh chỉ hợp lại ở bước SelectKBest trước khi PCA rút xuống ba chiều.
